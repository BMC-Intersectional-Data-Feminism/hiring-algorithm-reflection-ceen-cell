Now we're going to write algorithms to select applicants based on a variety of decisions. For each one, take note of how many applicants are passed onto the next stage of the application process.
```python
selection = [] ##1. create a list to hold the selected applicants

for applicant in applications:
    if applicant[6] > 80:
        selection.append(applicant)
        #print(selection)

print(selection)
```

Now let's try to create a filter for selecting applicants that have at least five grades above 85. **Hint**: A counter can be useful here! 

```python

selection_atleast_5gradesabove85 = [] ## create a list to hold the selected applicants with five grades above 85

for applicant in applications:
    count = 0
    for grade in applicant:
        if grade > 85:
            count+=1
        if count >= 5:
            selection_atleast_5gradesabove85.append(applicant)
            #print(selection_atleast_5gradesabove85)

print(selection_atleast_5gradesabove85)
```

```python
len(selection_atleast_5gradesabove85)
```

Finally, let's write a algorithm that selects applicants if they have an average grade of at least 85 across their six classes. 

```python
selection_avggradesabove85 = [] ## create a list to hold the selected applicants with five grades above 85


for applicant in applications:
    total = 0  # Reset sum for each applicant
    for grade in applicant :
        total += grade  # Sum up grades

    if total / 7 >= 85:  # Check if the average is at least 85
        selection_avggradesabove85.append(applicant)  # Append to the list

print(selection_avggradesabove85)
```

```python
print(len(selection_avggradesabove85))
```

While our data seemed to perfectly reported and without any inconsitencies, the world is less perfect. Consider the following scenarios: 

___
*Story 1*: Misread the Instructions
What if an excellent applicant thinks they should put in letter grades?

`[‘A’, ‘A’, ‘A’, ‘A’, ‘A’, ‘A’, ‘A’]`

. . . or how about their grades on 4-point scale?

`[4, 3.9, 4, 4, 3.95, 4, 3.9]`

___
*Story 2*: Bad Assumptions
What if one of your applicants skipped Intro to Computer Science? When they saw your form, they froze, and decided that putting -1 in the input field would make it obvious. . .

`[-1, 95, 99, 94, 96, 98, 95]`

___
*Story 3*: Mistake in the Input
What if one of your applicants accidentally put in a number > 100?

`[681, 68, 73, 70, 81, 91, 59]`

That might seem easy enough for a program to catch, but what if they accidentally dropped a 0?

`[100, 100, 100, 100, 100, 100, 10]`

A person would catch that mistake easily, does your algorithm?

___
*Story 4*: The Awful Semester
What if your applicant had a medical emergency one semester? Or a personal tragedy?

`[95, 93, 50, 91, 98, 90, 90]`

___
*Story 5*: Inverse Trajectories
What if one of your applicants came from an underprivileged background and really struggled at the beginning of college. . . but showed extraordinary growth by the end?

`[65, 75, 85, 95, 100, 100, 80]`

What if one of your applicants came to college with extraordinary potential? They easily aced their first few classes and then gradually grew apathetic about their education - getting nothing but barely-passing grades by the time they were a senior?

`[100, 100, 95, 85, 75, 65, 80]`

Does your algorithm treat them equally?

3. This algorithm amplifies the advantages of students from privileged backgrounds. Those with consistent access to resources, stable family lives, and quality education are more likely to have consistently high grades. This algorithm disadvantages students facing systemic challenges like poverty, discrimination, or personal hardship. These students may have inconsistent grades due to factors outside their control, even if they possess great potential. In Story 4, The Awful Semester and Story 5,  Inverse Trajectories, a single bad semester or an upward trajectory of improvement isn't adequately considered. This further marginalizes students who may have overcome significant obstacles to achieve academic success. Furthermore, the algorithm's rigidity, as seen in Story 1, Misread the Instructions, penalizes those unfamiliar with grading conventions in the US if they are international applicants or any local student who misread it.

4.To design a fair algorithm means to have an algorithm that minimizes bias and provides an equal grounds for all applicants to be assessed.There should be an understanding of the limitations of using grades as the sole measure of potential. Grades are a product of a complex system and do not fully capture an individual's abilities or potential. Potential biases in the data itself, like the historical inequities influencing access to quality education must be addressed
The algorithm's decision-making process should also be made transparent and understandable to applicants. This allows for scrutiny and identification of potential biases.


5. Including additional data like extracurriculars, internships, and recommendations can provide a more holistic view of the applicant. However, this can also introduce new biases. Students from privileged backgrounds may have greater access to prestigious extracurricular activities and internships, further reinforcing existing advantages.
Letters of recommendation can be biased, consciously or unconsciously, reflecting societal prejudices based on race, gender, or socioeconomic status.
To mitigate these biases, we can introduce a Standardized Evaluation. Using standardized rubrics to evaluate qualitative data like extracurriculars and recommendations, will help to minimize subjective interpretations. Also, we can incorporate a qualitative review process where humans consider the context and nuances of each application, especially in cases where the algorithm flags potential discrepancies or exceptional circumstances.


6. Current hiring filter algorithms, often used by large companies, analyze resumes and applications for keywords, skills, and experience. They encounter problems such as:
Bias Amplification. They akso often fail to capture the nuances of human experience and potential, overlooking non-traditional career paths or skills gained through unconventional means.Applicants can also learn to "game" the system by putting on their resumes keywords, even if they lack genuine expertise in those areas.
These algorithms share similarities with the admissions algorithm discussed in that they rely on quantifiable data and can lead to systemic inequalities. However, they often incorporate more diverse data points, including skills and work history. The key takeaway is that any algorithm used in high-stakes decision-making must be carefully designed and evaluated for fairness and potential biases limited as much as possible.
